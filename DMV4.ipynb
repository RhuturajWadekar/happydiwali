```python
# Data Wrangling on Real Estate Market

# Step 1: Import Libraries

import pandas as pd

import numpy as np

from sklearn.preprocessing import LabelEncoder

# Step 2: Load Dataset and Clean Column Names

df = pd.read_csv("realestate_prices.csv")

df.columns = (

    df.columns.str.strip()

    .str.lower()

    .str.replace(" ", "_")

    .str.replace(r"[^a-zA-Z0-9_]", "", regex=True)

)

print("Dataset Loaded Successfully")

print(df.head())

print("\nDataset Shape:")

print(df.shape)

# Step 3: Handle Missing Values

print("\nMissing Values:")

print(df.isnull().sum())

# Fill numerical missing values with median

for col in df.select_dtypes(include=np.number).columns:

    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode

for col in df.select_dtypes(include="object").columns:

    if not df[col].mode().empty:

        df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values After Handling:")

print(df.isnull().sum())

# Step 4: Merge Additional Data (if available)

# No additional dataset is provided.

# Step 5: Convert Date Column

df["date"] = pd.to_datetime(df["date"], errors="coerce")

# Step 6: Filter Data

filtered_df = df[df["date"].dt.year == 2015].copy()

print("\nFiltered Data (Year 2015):")

print(filtered_df.head())

# Step 7: Encode Categorical Variables

le = LabelEncoder()

df["zipcode"] = le.fit_transform(df["zipcode"].astype(str))

print("\nCategorical Data Encoded")

print(df[["zipcode"]].head())

# Step 8: Aggregate Data

avg_price = df.groupby("bedrooms")["price"].mean()

print("\nAverage House Price by Bedrooms:")

print(avg_price)

# Step 9: Identify and Handle Outliers Using IQR

Q1 = df["price"].quantile(0.25)

Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR

upper = Q3 + 1.5 * IQR

print("\nLower Bound:", lower)

print("Upper Bound:", upper)

df = df[

    (df["price"] >= lower) &

    (df["price"] <= upper)

]

print("\nOutliers Removed Successfully")

# Step 10: Display Final Dataset

print("\nFinal Dataset Shape:")

print(df.shape)

print("\nFinal Cleaned Dataset:")

print(df.head())

print("\nData Wrangling Completed Successfully!")
```
